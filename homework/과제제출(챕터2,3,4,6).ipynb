```python
help(len)
help(str.rfind)
help(str.islower)
help(str.strip)
help(str.title)
```

len(obj, /)
- Return the number of items in a container.
- → 컨테이너 안에 들어 있는 항목의 개수를 반환한다.

rfind(self, sub[, start[, end]], /)
- Return the highest index in S where substring sub is found
- → 문자열 S에서 부분 문자열 sub가 발견되는 가장 큰 인덱스를 반환한다.
- such that sub is contained within S[start:end]
- → 검색 범위를 S[start:end]로 제한할 수 있다.
- Return -1 on failure.
- → 찾지 못하면 -1을 반환한다.
- start와 end는 슬라이싱 방식처럼 해석됨.

islower(self, /)
- Return True if the string is a lowercase string, False otherwise.
- → 문자열이 소문자 문자열이면 True, 아니면 False를 반환한다.
- A string is lowercase if all cased characters ...
- → 대소문자를 구분할 수 있는 모든 문자가 소문자이고, 그런 문자가 최소 1개 이상 있으면 소문자 문자열로 판단한다.

strip(self, chars=None, /)
- Return a copy of the string with leading and trailing whitespace removed.
- → 문자열의 앞쪽과 뒤쪽 공백을 제거한 복사본을 반환한다.

More specifically, words start with uppercased characters and all remaining cased characters have lower case.
- → 더 정확히 말하면, 각 단어는 대문자로 시작하고 나머지 대소문자 구분 문자는 모두 소문자가 된다.

If encoding or errors is specified...

→ 그렇지 않으면 object.__str__() 결과를 사용하고, 없으면 repr(object) 결과를 사용한다.
