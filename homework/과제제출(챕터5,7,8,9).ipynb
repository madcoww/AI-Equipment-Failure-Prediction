```python
import pandas as pd
from sqlalchemy import create_engine, URL
from getpass import getpass

# MySQL 접속 정보
url = URL.create(
    "mysql+pymysql",
    username="root",
    password=getpass("MySQL 비밀번호: "),
    host="localhost",
    port=3306,
    database="data_test"
)

# MySQL 연결
engine = create_engine(url)

# 실제 테이블 이름으로 앞의 5행 조회
df = pd.read_sql(
    "SELECT * FROM molding_import_01 LIMIT 5",
    engine
)

print(df)
```
