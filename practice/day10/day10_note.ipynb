```python
import csv

fails = []

with open("../data/08_press.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    header = next(reader)

    for row in reader:
        if len(row) > 5 and row[5] == "1":
            fails.append(row)

with open("08_fail.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(header)

    for row in fails:
        writer.writerow(row)
```
