```python
with open("../data/09_ict_inspection.csv", "r", encoding="utf-8") as f:
    rows = f.readlines()
total = 0.0
skipped = 0
for line in rows[1:]:                 # 헤더 제외
    cols = line.strip().split(",")
    try:
        total += float(cols[2])       # 측정값(2열)
    except (ValueError, IndexError):
        skipped += 1
        continue                      # 불량 줄은 건너뜀
print("합계:", round(total, 1), "건너뜀:", skipped)
```

```python
v=5.0
```

```python
if not (0.0 <= v <= 5000.0):
    raise ValueError("정상 범위 밖")   # 우리 규칙을 벗어나면 직접 예외 발생
print(v)
```

```python
def check_value(text, lo=0.0, hi=5000.0):
    v = float(text)                       # 실패 시 ValueError
    if not (lo <= v <= hi):
        raise ValueError(f"정상 범위 밖: {v}")
    return v
```

```python
for t in ["128.5", "결측", "-999"]:
    try:
        print("정상값:", check_value(t))
    except ValueError as e:
        print("걸러짐:", e)
```
