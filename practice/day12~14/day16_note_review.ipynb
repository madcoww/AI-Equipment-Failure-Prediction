```python
from getpass import getpass
conn=pymysql.connect(
    host='localhost',
    port=3306,
    user='root',
    password=getpass('MySQL 비밀번호: '),
    database='shopdb'
)
```

```python
conn
```

```python
try:
    with conn.cursor() as cursor:
        sql='select * from membertbl'
        cursor.execute(sql)
        result=cursor.fetchall()
        for row in result:
            print(row)
except:
    print("db 접속시 문제발생")
```
