```python
John Smith	Intern	25000
Alex Du Bois	Team Lead	75000
Joanne Rowling	Manager	90000
Jane Connor	Manager	70000
```

```python
chart1 = [{"name":"Harry Potter", "Gender" : "Male", "age" : 23},
        {"name":"David Baker", "Gender" : "Male", "age" : 31},
        {"name":"John Smith", "Gender" : "Male", "age" : 22},
        {"name":"Juan Martinez", "Gender" : "Male", "age" : 36},
        {"name":"Jane Connor", "Gender" : "Female", "age" : 30}]

print(*chart1, sep="\n")
```

```python
chart2 = [{"name":"John Smith", "Position":"Intern", "wage":25000 },
        {"name":"Alex Du Bois", "Position":"Team Lead", "wage":75000 },
        {"name":"Joanne Rowling", "Position":"Manager", "wage":90000 },
        {"name":"Jane Connor", "Position":"Manager", "wage":70000 }]

print(*chart2, sep="\n")
```

```python
df = pd.DataFrame(chart1)
df
```

```python
df2 = pd.DataFrame(chart2)
df2
```

```python
merged_df = pd.merge(df, df2, on="name",how="inner")
merged_df
```

```python
merged_df = pd.merge(df, df2, on="name",how="left")
merged_df
```

```python
merged_df = pd.merge(df, df2, on="name",how="right")
merged_df
```

```python
merged_df = pd.merge(df, df2, on="name",how="outer")
merged_df
```
