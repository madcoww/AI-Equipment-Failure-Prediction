```python
df = pd.read_csv('C:\\AI Equipment Failure Prediction\\data\\14_hydraulic.csv')
```

```python
df.head()
```

```python
print(round(df['온도'].corr(df['진동']),3))
print(round(df['온도'].corr(df['압력']),3))
print(round(df['온도'].corr(df['냉각효율']),3))

```

```python
#std=표준편차 cov 공분산
```

```python
온도표준편차=round(df['온도'].std(),3)
print(round(온도표준편차,3))
```

```python
진동표준편차=round(df['진동'].std(),3)
print(round(진동표준편차,3))
```

```python
진동온도공분산=df['진동'].cov(df['온도'])
print(round(진동온도공분산,3))
```

```python
진동온도상관계수=진동온도공분산/(진동표준편차*온도표준편차)
print(round(진동온도상관계수,3))
```

```python
압력표준편차=round(df['압력'].std(),3)
print(round(압력표준편차,3))
```

```python
압력온도공분산=df['압력'].cov(df['온도'])
print(round(압력온도공분산,3))
```

```python
압력온도상관계수=압력온도공분산/(압력표준편차*온도표준편차)
print(round(압력온도상관계수,3))
```

```python
냉각효율표준편차=round(df['냉각효율'].std(),3)
print(round(냉각효율표준편차,3))
```

```python
냉각효율온도공분산=df['냉각효율'].cov(df['온도'])
print(round(냉각효율온도공분산,3))
```

```python
냉각효율온도상관계수=냉각효율온도공분산/(냉각효율표준편차*온도표준편차)
print(round(냉각효율온도상관계수,3))
```
