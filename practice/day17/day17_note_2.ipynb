```python
import pandas as pd

price = pd.read_excel('C:/AI Equipment Failure Prediction/data/Stock Data/stock price.xlsx')
valuation = pd.read_excel('C:/AI Equipment Failure Prediction/data/Stock Data/stock valuation.xlsx')
```

```python
price.head()  #name, stock name  겹침 수정함
```

```python
valuation.head()
```

```python
merged_prival = price.merge(valuation, on=['id', 'name'], how='outer').drop_duplicates()
```

```python
merged_prival.round(2).head(100)
```

```python
merged_prival[['name', 'price']]
```

```python
# 계산한 주가만 별도 변수에 저장
calculated_price = merged_prival['eps'] * merged_prival['per']

# 기존 주가의 NaN을 채운 결과를 새 변수에 저장
filled_price = merged_prival['price'].fillna(calculated_price)

# NaN이 채워진 주가를 기존 데이터프레임에 반영
merged_prival['price'] = filled_price
```

```python
# 원본을 유지하도록 표 복사
filled_df = merged_prival.copy()

# price의 결측값을 EPS × PER로 채우기
filled_df['price'] = filled_df['price'].fillna(
    filled_df['eps'] * filled_df['per']
)

# 전체 표 보기
display(filled_df)
```

| 컬럼 | 의미 |
|---|---|
| `name` | 회사명·종목명 |
| `value` | 회사가 돈 버는 능력(EPS)과 가진 순자산(BPS)을 함께 보고 주식 가치를 추정한다|
| `price` | 주가 |
| `eps` | 주당순이익 — 주식 1주당 순이익 |
| `bps` | 주당순자산 — 주식 1주당 순자산 |
| `per` | 주가수익비율 — **주가 ÷ EPS** |
| `pbr` | 주가순자산비율 — **주가 ÷ BPS** |

```python
filled_df['value'] = filled_df['value'].fillna(
    (filled_df['eps'] * 10 + filled_df['bps']) / 2
)

display(filled_df)
```
