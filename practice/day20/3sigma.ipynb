```python
import os

import numpy as np


def dp(name):
    """실행 위치가 어디든 data 폴더를 찾아 주는 도우미"""
    here = os.path.dirname(os.path.abspath(__file__))
    for c in (os.path.join("data", name),
              os.path.join(here, "data", name),
              os.path.join(here, "..", "data", name)):
        if os.path.exists(c):
            return c
    return os.path.join("data", name)
```

```python
Fcc = 'C:\\AI Equipment Failure Prediction\\data\\11_hydraulic_cooler.csv'
```

```python
df=pd.read_csv(Fcc)
```

```python
df.describe()
```

```python
#온도의3시그마상한을구하고그위에있는사이클번호를출력
temp = df['temp']
upper_limit = temp.mean() + 3 * temp.std()
cycles_above_limit = df.loc[temp > upper_limit, 'cycle']
print(cycles_above_limit)
```

```python
#2.온도의 평균과 흩어진 정도를 구하기
temp = df['temp']
mean_temp = temp.mean()
std_temp = temp.std()

print("평균 온도:", mean_temp)
print("표준 편차:", std_temp)
```

```python
#3. 평균에 흩어진 정도의 세 배를 더해 상한 만들기
upper_limit = mean_temp + 3 * std_temp
print("3시그마 상한:", upper_limit)
```

```python
cycles_above_limit = df.loc[temp > upper_limit, 'cycle']
print("상한을 넘는 사이클 번호:", cycles_above_limit)
print("총 상한을 넘는 사이클 수:", len(cycles_above_limit))
print("상한을 넘는 사이클 비율:", len(cycles_above_limit) / len(df))
```

```python
#           5. 상한값과 걸린 사이클 번호를 함께 출력해 실습 3 결과와 비교

for cycle in cycles_above_limit:
    temp_value = df.loc[df['cycle'] == cycle, 'temp'].values[0]
    print(f"사이클 {cycle}: 상한값 {upper_limit}, 실제 온도 {temp_value}")
    if temp_value > upper_limit:
        print(f"사이클 {cycle}는 상한을 초과했습니다.")
    else:
        print(f"사이클 {cycle}는 상한을 초과하지 않았습니다.")
```
