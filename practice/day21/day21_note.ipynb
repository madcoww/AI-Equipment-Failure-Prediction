```python
df = pd.read_csv(heat)
```

```python
df.head()
```

```python
df['배치'].unique() # 배치 컬럼의 고유값 확인
```

```python
a=df[df['배치'] == '로트A']
```

```python
plt.plot(a['측정시각'], a['온도'])
plt.show()
```

```python
plt.figure(figsize=(10,4)) 
plt.plot(a['측정시각'], a['온도'], marker='o', linestyle='-')
plt.xlabel('측정시각')
plt.ylabel('온도')
plt.grid(True)
plt.legend(['온도'])
plt.title('로트A 온도 변화')
plt.show()
```

```python
b=df[df['배치'] == '로트B']
c=df[df['배치'] == '로트C']
```

```python
plt.figure(figsize=(10,4)) 
plt.plot(a['측정시각'], a['온도'])
plt.plot(b['측정시각'], b['온도'])
plt.plot(c['측정시각'], c['온도'])
plt.xlabel('측정시각')
plt.ylabel('온도')
plt.legend(['로트A', '로트B', '로트C'])
plt.title('로트별 온도 변화')
plt.show()
```

```python
plt.figure(figsize=(10,4)) 
plt.plot(a['측정시각'], a['온도'], label='로트A')
plt.plot(b['측정시각'], b['온도'], label='로트B')
plt.plot(c['측정시각'], c['온도'], label='로트C')
plt.xlabel('측정시각')
plt.ylabel('온도')
plt.grid(True, alpha=0.3)
plt.legend(loc='upper left')
plt.title('로트별 온도 변화')
plt.show()
```

```python
train = ("C:\\AI Equipment Failure Prediction\\data\\결측치 잇는데이터\\train.csv")
dft = pd.read_csv(train)
```

```python
dft.head()
```

```python
일시=dft['일시'].head(50)
```

```python
전력=dft['전력소비량(kWh)'].head(50)
```

```python
기온=dft['기온(C)'].head(50)
```

```python
plt.figure(figsize=(12,4)) 
plt.title('전력 소비량 변화')
plt.xlabel('일시')
plt.xticks([])
plt.ylabel('전력소비량(kWh)')
plt.plot(일시,전력)
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12,4)) 
plt.title('기온 변화')
plt.xticks([])
plt.xlabel('일시')
plt.ylabel('기온(C)')
plt.grid(True)
plt.plot(일시,기온, c='red')
plt.show()
```

```python
plt.figure(figsize=(12, 4))
plt.title('전력소비량과 기온 변화')
plt.xlabel('일시')
plt.xticks([])

plt.ylabel('전력소비량(kWh)')
plt.plot(일시, 전력, c='blue')
plt.grid(True)

plt.twinx()
plt.ylabel('기온(C)')
plt.plot(일시, 기온, c='red')

plt.show()
```

```python
plt.scatter(일시, 전력, c='blue')
plt.xlabel('일시')
plt.xticks([])
plt.ylabel('전력소비량(kWh)')
plt.title('전력소비량 변화')
plt.grid(True)
plt.show()
```
