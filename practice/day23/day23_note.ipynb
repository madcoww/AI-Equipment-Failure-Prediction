```python
plt.figure(figsize=(10,6))
sns.scatterplot(df, x='소입로온도', y='건조출력', hue='판정', alpha=0.6)
plt.title('소입로온도 vs 건조출력')
plt.xlabel('소입로온도')
plt.ylabel('건조출력')
plt.show()

#해석 
# 소입로온도가 높아지면 건조출력도 같이 올라가는 것 같습니다.
```

```python
df['소입로온도'].corr(df['건조출력'])
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(df, x='소입로온도', y='제어출력', hue='판정', alpha=0.6)
plt.title('소입로온도 vs 제어출력')
plt.xlabel('소입로온도')
plt.ylabel('제어출력')
plt.show()
```

```python
df['소입로온도'].corr(df['제어출력'])
```

```python
df.corr(numeric_only=True).round(2)
```

```python
sns.set_theme(style="whitegrid", font="Malgun Gothic")
corr=df.corr(numeric_only=True).round(2)
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True,fmt=".2f", cmap="coolwarm", center=0)
plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False
plt.title('상관 히트맵')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12,5))
sns.histplot(data=df, x="온도 ", ax=ax1)
sns.boxplot(data=df, x="설비라인 ", y="진동 ", ax=ax2)
plt.tight_layout()
```
