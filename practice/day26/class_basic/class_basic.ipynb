#파이선 클래스간의 관계 
1. 상속(Inheritance)
2. 포함(Composition)
3. 연관(Association)
#상속(Inheritance): 기존 클래스의 기능을 재사용하여 새로운 클래스를 만드는 것
#포함(Composition): 한 클래스가 다른 클래스를 포함하여 기능을 확장하는 것
#연관(Association): 한 클래스가 다른 클래스와 관계를 맺는 것

```python
# 상속  
```

```python
class Animal : 
    def describe(self):
        print('동물이다!')
```

```python
class Dog(Animal):
    def describe(self):
        print('강아지다!')
```

```python
class Dog(Animal):
    pass
```

```python
pupuy = Dog()
```

```python
pupuy.describe()
```
