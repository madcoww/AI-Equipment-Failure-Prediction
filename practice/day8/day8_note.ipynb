```python
A={1,3,5}  #A사건 홀수 
B={1,2,3}  #B사건 3이하
print(A.intersection(B)) # 교집합(intersection) : 두 집합에 공통으로 있는 값
print(A.union(B))   # 합집합(union) : 두 집합의 모든 값을 합친 것
print(A.difference(B))  #차집합(difference) : A에는 있지만 B에는 없는 값
```
