```python
#4 셋으로중복센서제거하기
logs = ["S01", "S02", "S01", "S03", "S02"]
```

```python
unique = set(logs)
print(sorted(unique))
print(len(unique))
```

```python
#5 두라인의센서구성비교하기
line_a = {"S01", "S02", "S03", "S05"}
line_b = {"S03", "S04", "S05"}

print(line_a.union(line_b))
print(line_a.intersection(line_b))
print(line_a.difference(line_b))
print(line_b.difference(line_a))



```
