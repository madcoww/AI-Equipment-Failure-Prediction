```python
import numpy as np
import pandas as pd
import tensorflow as tf
    
```

```python
df=pd.read_csv("C:\\AI Equipment Failure Prediction\\tenserflow\\titanic.csv")
```

```python
df.shape
```

```python
df.isna().sum()
```

```python
df = df.dropna()
df.isna().sum()
```

```python
model = tf.keras.models.Sequential([        #keras란? 딥러닝 모델을 쉽게 구성하고 학습시킬 수 있는 고수준 API
    tf.keras.layers.Dense(64, activation='tanh'),
    tf.keras.layers.Dense(32, activation='tanh'),  #relu 외 다른 모델   예: tanh, sigmoid 등
    # sigmoid: 출력을 0~1 사이로 바꿔 이진 분류의 생존 확률 추정값으로 사용한다.
    tf.keras.layers.Dense(1, activation='sigmoid')
])
```

```python
# compile: 평가 기준을 설정한다.
# loss='binary_crossentropy': 생존 0/1처럼 두 범주를 구분할 때 쓰는 손실 함수다.
# metrics=['accuracy']: 학습 중 정답 비율을 표시한다. 손실 함수와 별도의 평가 지표다.
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy']) 
```

```python
df.columns.tolist()
```

```python
x = df[['운임', '성별', '객실등급']]
x.head()
```

```python
y = df['생존여부']
```

```python
model.fit(x, y, epochs=500)  #fit 줄임말 X, 맞추다.
```

```python
# accuracy: 정답률 - loss: 손실함수
# 딥러닝 모델 평가
```
